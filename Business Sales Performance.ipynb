```python
## Category-wise Sales & Profit

category_sales_profit = df.groupby('Category')[['Sales', 'Profit']].sum()
category_sales_profit
```

```python
## Region-wise Sales & Profit

region_sales_profit = df.groupby('Region')[['Sales', 'Profit']].sum()
region_sales_profit
```

```python
## Segment-wise Sales & Profit

segment_sales_profit = df.groupby('Segment')[['Sales', 'Profit']].sum()
segment_sales_profit
```

```python
## Profit Margin(%) 

category_sales_profit['Profit Margin %'] = (category_sales_profit['Profit'] / category_sales_profit['Sales']) * 100
category_sales_profit
```

```python
## Region: 


region_sales_profit['Profit Margin %'] = (region_sales_profit['Profit'] / region_sales_profit['Sales']) * 100
region_sales_profit
```

```python
## Segment:

segment_sales_profit['Profit Margin %'] = (segment_sales_profit['Profit'] / segment_sales_profit['Sales']) * 100
segment_sales_profit
```

```python
## Sub-Category Analysis

subcategory_sales_profit = df.groupby('Sub-Category')[['Sales', 'Profit']].sum()

subcategory_sales_profit.sort_values(
    by='Profit',
    ascending=False
)
```

```python
## Discount vs Profit Analysis 

discount_profit = df.groupby('Discount')['Profit'].sum()
discount_profit
```

```python
df[['Discount', 'Profit']].corr()
```

```python
## Year-wise Sales & Profit

yearly_sales_profit = df.groupby('Year')[['Sales', 'Profit']].sum()
yearly_sales_profit
```

```python
yearly_sales_profit = df.groupby('Year')[['Sales', 'Profit']].sum()

yearly_sales_profit['Sales_Growth'] = yearly_sales_profit['Sales'].pct_change() * 100
yearly_sales_profit['Profit_Growth'] = yearly_sales_profit['Profit'].pct_change() * 100

yearly_sales_profit
```

```python
## Conclusion

2014–2017 ke period mein overall Sales aur Profit dono grow hue. 
2016 mein Sales aur Profit ki growth sabse strong rahi, 
jabki 2015 mein sales decline ke bawajood profit significantly improve hua. 
```

```python
## Sales aur Profit ka relationship check 

yearly_sales_profit[['Sales', 'Profit']].corr()
```

```python
## Conclusion

Sales and Profit have a strong positive relationship, with a correlation coefficient of 0.94. 
This indicates that higher sales are generally associated with higher profit.
```

```python
## Final Year-wise Sales & Profit Chart
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

plt.plot(
    yearly_sales_profit.index,
    yearly_sales_profit['Sales'],
    marker='o',
    linewidth=2,
    label='Sales'
)

plt.plot(
    yearly_sales_profit.index,
    yearly_sales_profit['Profit'],
    marker='o',
    linewidth=2,
    label='Profit'
)

# Values show karna
for year, value in zip(yearly_sales_profit.index, yearly_sales_profit['Sales']):
    plt.text(year, value, f'{value:,.0f}', ha='center', va='bottom')

for year, value in zip(yearly_sales_profit.index, yearly_sales_profit['Profit']):
    plt.text(year, value, f'{value:,.0f}', ha='center', va='bottom')

plt.title('Year-wise Sales and Profit', fontsize=16, fontweight='bold')
plt.xlabel('Year')
plt.ylabel('Amount')
plt.xticks(yearly_sales_profit.index)
plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
## Final Business Insights

* Sales increased from **484,247.50 in 2014** to **733,215.26 in 2017**.
* Profit increased from **49,543.97 in 2014** to **93,439.27 in 2017**.
* **2016** recorded the strongest growth, with **29.47% Sales Growth** and **32.74% Profit Growth**.
* In **2015**, Sales decreased by **2.83%**, while Profit increased by **24.37%**.
* Sales and Profit have a **strong positive correlation of 0.9414**, indicating that higher sales were generally associated with higher profit.
```

```python
## Conclusion

Overall, the business showed a positive growth trend from **2014 to 2017**, with **2017 achieving the highest Sales and Profit**.
```

```python
## Category-wise Sales & Profit Analysis

* **Technology** generated the highest Sales (**836,154.03**) and the highest Profit (**145,454.95**).
* **Office Supplies** generated Sales of **719,047.03** and Profit of **122,490.80**.
* **Furniture** generated relatively high Sales (**741,999.80**) but comparatively low Profit (**18,451.27**).
* Overall, **Technology is the best-performing category**, while **Furniture requires further analysis** because of its low profit despite relatively high sales.
```

```python
## Sub-Category Analysis

* **Copiers** generated the highest profit of **55,617.82**, followed by **Phones** and **Accessories**.
* **Tables** recorded the highest loss of **17,725.48**, despite generating relatively high sales.
* **Bookcases** and **Supplies** also recorded negative profits.
* The analysis shows that **high sales do not always result in high profit**.
* Overall, **Copiers is the most profitable sub-category, while Tables requires further investigation due to its significant loss**.
```

```python
##  Discount vs Profit Analysis

* The highest total profit was recorded at **0% discount**, with a profit of **320,987.60**.
* Profit was relatively low at **10% and 15% discount** levels.
* At higher discount levels such as **30%, 40%, 50%, 60%, 70%, and 80%**, the business recorded overall losses.
* The **70% discount level recorded the highest loss of 40,075.36**.
* Overall, higher discount levels are generally associated with **lower profitability**, although the relationship is not perfectly linear.
```

```python
## Region-wise Sales & Profit Analysis

* **West** is the best-performing region, with the highest Sales of **725,457.82**, Profit of **108,418.45**, and Profit Margin of **14.94%**.
* **East** is the second-highest region in terms of Sales and Profit.
* **South** has the lowest Sales but maintains a healthy Profit Margin of **11.93%**.
* **Central** has the lowest Profit Margin of **7.92%**, indicating a need for further investigation.
* Overall, **West shows the strongest regional performance**, while **Central requires attention to improve profitability**.
```

```python
## State-wise Sales & Profit Analysis

state_sales_profit = df.groupby('State')[['Sales', 'Profit']].sum()

state_sales_profit.sort_values(by='Profit',ascending=False)
```

```python
##  State-wise Sales & Profit Analysis

* **California** is the top-performing state, generating the highest Sales of **457,687.63** and Profit of **76,381.39**.
* **New York** is the second-highest state in terms of Profit, with **74,038.55**.
* **Washington** also shows strong profitability with a Profit of **33,402.65**.
* Several states recorded negative profits, including **Texas, Ohio, Pennsylvania, Illinois, and North Carolina**.
* **Texas recorded the highest loss of 25,729.36**, despite generating Sales of **170,188.05**.
* Overall, the analysis shows that **high Sales do not always result in high Profit**, making profitability an important measure for evaluating state-level performance.
```

```python
## Segment-wise Sales & Profit Analysis

* **Consumer** is the best-performing segment, generating the highest Sales of **1,161,401.00** and Profit of **134,119.21**.
* **Corporate** generated Sales of **706,146.40** and Profit of **91,979.13**.
* **Home Office** generated the lowest Sales of **429,653.10** and Profit of **60,298.68** among the three segments.
* Overall, the **Consumer segment contributes the highest Sales and Profit**, making it the strongest customer segment.
```

```python
shipmode_sales_profit = df.groupby('Ship Mode')[['Sales', 'Profit']].sum()

shipmode_sales_profit.sort_values(
    by='Profit',
    ascending=False
)
```

```python
## Ship Mode-wise Sales & Profit Analysis

* **Standard Class** generated the highest Sales of **1,358,216.00** and Profit of **164,088.79**.
* **Second Class** ranked second, with Sales of **459,193.60** and Profit of **57,446.64**.
* **First Class** generated Sales of **351,428.40** and Profit of **48,969.84**.
* **Same Day** had the lowest Sales of **128,363.10** and Profit of **15,891.76**.
* Overall, **Standard Class is the dominant shipping mode in terms of total Sales and Profit**.
```

```python
## Top 10 Products by Profit

top_products_profit = (
    df.groupby('Product Name')['Profit']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_products_profit
```

```python
## Month-wise Sales & Profit

df['Order Date'] = pd.to_datetime(df['Order Date'])

monthly_sales_profit = (
    df.groupby(df['Order Date'].dt.to_period('M'))[['Sales', 'Profit']]
    .sum()
)

monthly_sales_profit
```

```python
## Monthly Sales & Profit Analysis

* **November 2017** recorded the highest monthly Sales of **118,447.83**.
* **February 2014** recorded the lowest monthly Sales of **4,519.89**.
* **December 2016** recorded the highest monthly Profit of **17,885.31**.
* Losses were recorded in **July 2014** and **January 2015**, with January 2015 showing the lowest Profit of **-3,281.01**.
* Monthly Sales and Profit show fluctuations throughout the four-year period, with **strong Sales levels observed toward the end of 2017**.
```

```python
## Monthly Sales & Profit Trend Chart

import matplotlib.pyplot as plt

plt.figure(figsize=(14, 6))

plt.plot(
    monthly_sales_profit.index.astype(str),
    monthly_sales_profit['Sales'],
    marker='o',
    label='Sales'
)

plt.plot(
    monthly_sales_profit.index.astype(str),
    monthly_sales_profit['Profit'],
    marker='o',
    label='Profit'
)

plt.title('Monthly Sales and Profit Trend', fontsize=16, fontweight='bold')
plt.xlabel('Month')
plt.ylabel('Amount')
plt.legend()
plt.grid(True, alpha=0.3)

# Har 3rd month ka label
plt.xticks(
    range(0, len(monthly_sales_profit), 3),
    monthly_sales_profit.index.astype(str)[::3],
    rotation=45
)

plt.tight_layout()
plt.show()
```
